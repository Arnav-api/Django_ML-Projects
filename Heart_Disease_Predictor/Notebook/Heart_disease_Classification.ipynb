```python
# Select numerical columns to scale
num_cols = ['Age', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR']

# Initialize and fit scaler on ALL at once
sc = StandardScaler()
data[num_cols] = sc.fit_transform(data[num_cols])

# Save the fitted scaler
```

```python
data.head()
```

```python
ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')
category = ohe.fit_transform(data[['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope']])
category
```

```python
category.drop(['Sex_M','ChestPainType_ASY','RestingECG_ST','ExerciseAngina_N','ST_Slope_Flat'],axis=1,inplace=True)
```

```python
category
```

```python
data = pd.concat([data,category],axis = 1).drop(['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope'],axis = 1)
```

```python
data.head()
```

```python
data_fit = data.drop('HeartDisease',axis=1)
data_fit.head()
```

```python
review = StratifiedKFold(n_splits=9)
def get_score(model, x_train , x_test , y_train , y_test):
    model.fit(x_train,y_train)
    return model.score(x_test,y_test)
```

```python
for train_index , test_index in review.split(data_fit,data['HeartDisease']):
    x_train, x_test = data_fit.iloc[train_index], data_fit.iloc[test_index]
    y_train, y_test = data['HeartDisease'].iloc[train_index], data['HeartDisease'].iloc[test_index]
    print(get_score(RandomForestClassifier(), x_train, x_test, y_train, y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Define parameter grid
param_grid = {
    'n_estimators': [200,300],           # number of trees
    'max_depth': [ 10, 20],           # depth of each tree
    'min_samples_split': [5, 10],           # minimum samples to split a node
    'min_samples_leaf': [2, 4],             # minimum samples at a leaf node
                    # whether to bootstrap samples
}

# Create model
rf = RandomForestClassifier(random_state=42)

# GridSearchCV setup
grid_search = GridSearchCV(estimator=rf,
                           param_grid=param_grid,
                           cv=4,                      # 5-fold CV
                           scoring='accuracy',        # or 'f1', 'roc_auc' etc.
                           n_jobs=-1,                 # use all cores
                           verbose=1)

# Fit the model on training data
grid_search.fit(x_train, y_train)
best_model = grid_search.best_estimator_

```

```python
model_score = best_model.score(x_test,y_test)
```

```python
model_score
```

```python
model = best_model
model
```

```python
# Save model
joblib.dump(model, 'model.pkl')

# Save scaler
joblib.dump(sc, 'scaler.pkl')

# Save encoder
joblib.dump(ohe, 'encoder.pkl')
```
